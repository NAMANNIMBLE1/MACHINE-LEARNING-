# ANN IMPLEMENTATION

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Flatten , Input
```

```python
(X_train , y_train) , (X_test , y_test) = keras.datasets.mnist.load_data()
```

```python
# values in range 0 - 255 but we want it to range 0 -> 1 ( sijmilar to noramlization)
X_train = X_train/255
X_test = X_test/255
```

```python
X_train[0]
```

```python
plt.imshow(X_train[0])
```

```python
# first step for neural network is to flatten ---> convert 2d datset to 1 d
model = Sequential()
model.add(Input(shape = (28 , 28))) # define input size of date 
model.add(Flatten())
model.add(Dense(128 , activation='relu'))
model.add(Dense(10 , activation= 'softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam' , loss='sparse_categorical_crossentropy' , metrics=['accuracy'])
```

```python
history = model.fit(X_train , y_train , epochs= 30 , validation_split= 0.1 , batch_size= 32)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
y_prob = model.predict(X_test)
```

```python
y_prob[0]
```

```python
y_pred = y_prob.argmax(axis = 1 )
```

```python
from sklearn.metrics import accuracy_score 

accuracy_score(y_pred , y_test)
```
